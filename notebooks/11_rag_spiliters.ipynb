```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = "This is a sample text that we will split into smaller chunks using the RecursiveCharacterTextSplitter. The purpose of this is to demonstrate how to effectively break down larger texts into manageable pieces for further processing or analysis."
```

```python
spilitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10,
)
```

```python
chunks = spilitter.split_text(text)
```

### DOCUMENT TEXT LOADER

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../data/AYUSH_230620.txt")
texts = loader.load()
```

```python
pdfLoader = PyPDFLoader("../data/AYUSH_230620.pdf")
pdfs = pdfLoader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
)
```

```python
text_chunks = splitter.split_documents(text)
```

```python
text_chunks
```

```python
pdfChunks = splitter.split_documents(pdfs)
```

```python
pdfChunks
```
