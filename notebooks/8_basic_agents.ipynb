```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(num1: int, num2: int):
    """
    Adds two numbers.
    args:
        num1 (int): The first number.
        num2 (int): The second number.
    """
    return num1 + num2

@tool 
def multiply_numbers(num1: int, num2: int):
    """
    Multiplies two numbers.
    args:
        num1 (int): The first number.
        num2 (int): The second number.
    """
    return num1 * num2
```

```python
add_numbers.invoke({"num1": 5, "num2": 3})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=1.0,
    max_retries=0,
)
```

```python
agent = create_agent(
    model = llm,
    tools = [add_numbers, multiply_numbers],
    system_prompt = "You are a helpful assistant that can perform basic arithmetic operations and always use tools for calculations.",
)
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "What is the sum of 10 + 20 * 10?"}]})
print(response["messages"][-1].content)
```
