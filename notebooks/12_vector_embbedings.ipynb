```python
documents = [
    "GenAI is the most important thing",
    "GenAI ussually use Python",
    "GenAI is new technology",
    "GenAI is used in many fields"
]
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
vector = embeddings.embed_documents(documents)
```

```python
len(vector)
```

### Storing vector embedding in chromadb


```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(documents, embeddings, persist_directory="./chroma_db") 
```

```python
query = "What is GenAI?"
results = vector_store.similarity_search(query, k=2 )
```
