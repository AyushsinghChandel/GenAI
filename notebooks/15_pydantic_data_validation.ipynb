```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class UserModel(BaseModel):
    name: str
    email: str
    age: int
```

```python
user = UserModel(name = "John Doe", email = "example@gmail.com", age = 30")
```

```python
print(user.age)
```

### FIELD VALIDATION

```python
class Product(BaseModel):
    name: str = Field(max_length=10)
    des: str = Field(min_length=25)
    price: float = Field(gt=0, le=100)
    

```

```python
Products = Product(name = "Product 2", des = "This is a product description", price = 50.0)
```

```python
class Order(BaseModel):
    id: str = Field(max_length=10, min_length=5)
    product:Product 
```

```python
order = Order(id = "12345", product = {"name": "Product 2", "des": "This is a product description", "price": 50.0})
```

```python
print(order.product.name)
```

```python
class Cart(BaseModel):
    id: str = Field(max_length=10, min_length=5)
    items: list[Product] = Field(min_items=0, max_items=5)
    comment: str = Field(max_length=100, default="No comment")

```

```python
cart = Cart(id = "12345", items = [{"name": "Product 2", "des": "This is a product description", "price": 50.0}, {"name": "Product 3", "des": "This is a product description", "price": 50.0}], comment = "This is a comment")
```
